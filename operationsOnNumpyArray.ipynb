```python
#  Create numpy array and perform following operations:
import numpy as np
# Convert 1D array to 2D
arr_1d = np.array([1, 2, 3, 4, 5, 6])
arr_2d = arr_1d.reshape(2, 3)
print("Original 1D array:\n", arr_1d)
print("Converted 2D array:\n", arr_2d)
# same could be done by using np.newaxis
arr_2d_newaxis = arr_1d[np.newaxis, :].reshape(2, 3)
print("Converted 2D array using np.newaxis:\n", arr_2d_newaxis)
# also same could be done by using np.expand_dims
arr_2d_expand_dims = np.expand_dims(arr_1d, axis=0).reshape(2, 3)
print("Converted 2D array using np.expand_dims:\n", arr_2d_expand_dims)
# same could be done by using np.atleast_2d
arr_2d_atleast_2d = np.atleast_2d(arr_1d).reshape(2, 3)
print("Converted 2D array using np.atleast_2d:\n", arr_2d_atleast_2d)
# same could be done by using np.array with ndmin parameter
arr_2d_ndmin = np.array(arr_1d, ndmin=2).reshape(2, 3)
print("Converted 2D array using np.array with ndmin parameter:\n", arr_2d_ndmin)
# same using vstack
arr_2d_vstack = np.vstack((arr_1d[:3], arr_1d[3:]))
print("Converted 2D array using np.vstack:\n", arr_2d_vstack)
```

```python
# Reverse a 1D NumPy array
arr = np.array([1, 2, 3, 4, 5])
reversed_arr = arr[::-1]
print("Original array:", arr)
print("Reversed array:", reversed_arr)
```

```python
# Create a 4×4×3 array and extract value at its second set, first row and last column
arr_3d = np.random.rand(4, 4, 3)
print("3D Array:\n", arr_3d)
extracted_value = arr_3d[1, 0, -1]
print("Extracted value at secon" \
"d set, first row and last column:", extracted_value)
```

```python
# Create a 4×4 and Extract Odd Rows and Even Columns 
arr_4x4 = np.random.rand(4, 4)
print("Original 4x4 Array:\n", arr_4x4)
odd_rows_even_cols = arr_4x4[1::2, ::2]
print("Extracted Odd Rows and Even Columns:\n", odd_rows_even_cols)
```

```python
# Slice the first two rows and first two columns of econd set from a 4×4×3 array
arr_3d = np.random.rand(4, 4, 3)
print("3D Array:\n", arr_3d)
sliced_arr = arr_3d[1, :2, :2]
print("Sliced array (first two rows and first two columns of second set):\n", sliced_arr)
```

```python
#  Replace all odd numbers in a NumPy array with -1 by itterating using for loop [[23, 56, 78, 93], [71, 82,13, 24]]
arr = np.array([[23, 56, 78, 93], [71, 82, 13, 24]])
print("Original array:\n", arr)
for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        if arr[i, j] % 2 != 0:
            arr[i, j] = -1
print("Array after replacing odd numbers with -1:\n", arr)
```

```python
# Get the indices of non-zero elements in an array [1, 0, 2, 0, 3, 0, 4]
arr = np.array([1, 0, 2, 0, 3, 0, 4])
non_zero_indices = np.nonzero(arr)
print("Indices of non-zero elements:", non_zero_indices)
```

```python
# Perform arithmetic operations on two NumPy arrays element-wise. Add two NumPy arrays element by element. Multiply two NumPy arrays element by element.
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
sum_arr = arr1 + arr2
diff_arr = arr1 - arr2
prod_arr = arr1 * arr2
quot_arr = arr1 / arr2
print("Element-wise sum:", sum_arr)
print("Element-wise difference:", diff_arr)
print("Element-wise product:", prod_arr)
print("Element-wise quotient:", quot_arr)
```
